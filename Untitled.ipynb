```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
```

```python
train_path = r"C:\Users\HP\OneDrive\Desktop\EMoBhool\fer213\train"
test_path = r"C:\Users\HP\OneDrive\Desktop\EMoBhool\fer213\test"

#checking if it true
print(f'Train folders: {os.listdir(train_path)}')
print(f'Test folders: {os.listdir(test_path)}')
```

```python
def load_data(data_path):
    X, y = [], []
    emotions = os.listdir(data_path)  # List all emotion categories

    for emotion in emotions:
        emotion_folder = os.path.join(data_path, emotion)
        label = emotions.index(emotion)  # Assign a label to each emotion

        for img_name in os.listdir(emotion_folder):
            img_path = os.path.join(emotion_folder, img_name)
            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)  # Read as grayscale
            img = cv2.resize(img, (48, 48))  # Resize to 48x48

            X.append(img)
            y.append(label)

    X = np.array(X).astype('float32') / 255.0  # Normalize the pixel values
    y = np.array(y)

    return X, y

# Load the train and test data
X_train, y_train = load_data(train_path)
X_test, y_test = load_data(test_path)

print(f'Train data shape: {X_train.shape}, Train labels shape: {y_train.shape}')
print(f'Test data shape: {X_test.shape}, Test labels shape: {y_test.shape}')
```
